```python
import torch
from transformers import pipeline

model_path = "cardiffnlp/twitter-roberta-base-sentiment-latest"

sentiment_task = pipeline(
    task="sentiment-analysis",
    model=model_path,
    tokenizer=model_path,
    device=-1
)

print(sentiment_task("Good morning"))

```
